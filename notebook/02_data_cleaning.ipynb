# 02 - Data Cleaning

```python
import pandas as pd

students = pd.read_csv("../data/raw/students.csv")
scores = pd.read_csv("../data/raw/scores.csv")
activity = pd.read_csv("../data/raw/activity.csv")
```

```python
students=students.drop_duplicates()
students.shape
```

```python
students["region"]=students["region"].str.strip().str.title()
students["region"].value_counts()
```

```python
students["highest_education"]=students["highest_education"].fillna("Unknown")
```

```python
students["highest_education"].value_counts()
```

```python
print(students.isna().sum())
students
```

```python
print(students.shape)
print(students.duplicated().sum())
```

```python
students.to_csv("../data/processed/students_clean.csv", index=False)
scores.to_csv("../data/processed/scores_clean.csv", index=False)
activity.to_csv("../data/processed/activity_clean.csv", index=False)
```
